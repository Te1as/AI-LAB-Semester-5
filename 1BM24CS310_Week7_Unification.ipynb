```python

def is_variable(x):
    return isinstance(x, str) and x[0].islower()


def occurs_check(var, expr):
    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs_check(var, arg) for arg in expr[1])

    return False


def substitute(expr, subst):
    if isinstance(expr, str):
        while expr in subst:
            expr = subst[expr]
        return expr

    if isinstance(expr, tuple):
        return (expr[0], tuple(substitute(arg, subst) for arg in expr[1]))

    return expr


def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    x = substitute(x, subst)
    y = substitute(y, subst)

    # Step 1: If both are identical
    if x == y:
        return subst

    # Step 2: If x is a variable
    if is_variable(x):
        if occurs_check(x, y):
            return None
        subst[x] = y
        return subst

    # Step 3: If y is a variable
    if is_variable(y):
        if occurs_check(y, x):
            return None
        subst[y] = x
        return subst

    # Step 4: Check predicate symbols
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        return None

    if x[0] != y[0]:
        return None

    # Step 5: Check number of arguments
    if len(x[1]) != len(y[1]):
        return None

    # Step 6: Unify corresponding arguments
    for a, b in zip(x[1], y[1]):
        subst = unify(a, b, subst)

        if subst is None:
            return None

    return subst


# Example
expr1 = ("Knows", ("x", "John"))
expr2 = ("Knows", ("Riya", "x"))

result = unify(expr1, expr2)

if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("SUBST =", result)
```
