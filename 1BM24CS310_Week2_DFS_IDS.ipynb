```python
def dfs(start, goal):
    stack = [(start, [])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path + [state]

        if state in visited:
            continue

        visited.add(state)
        zero = state.index(0)
        row, col = zero // 3, zero % 3

        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        for dr, dc in moves:
            nr, nc = row + dr, col + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                new_zero = nr * 3 + nc
                new_state = list(state)

                new_state[zero], new_state[new_zero] = (
                    new_state[new_zero], new_state[zero]
                )

                new_state = tuple(new_state)

                if new_state not in visited:
                    stack.append((new_state, path + [state]))

    return None


def display(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

solution = dfs(start, goal)

if solution:
    print("Solution found!")
    for step in solution:
        display(step)
    print("Total moves:", len(solution) - 1)
else:
    print("No solution found.")
```

```python
def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row, col = zero // 3, zero % 3

    # Up, Down, Left, Right
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        nr, nc = row + dr, col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_zero = nr * 3 + nc
            new_state = list(state)

            new_state[zero], new_state[new_zero] = (
                new_state[new_zero], new_state[zero]
            )

            neighbors.append(tuple(new_state))

    return neighbors


def depth_limited_dfs(state, goal, limit, path, path_set):
    # Goal found
    if state == goal:
        return path[:]

    # Stop if the depth limit is reached
    if len(path) - 1 == limit:
        return None

    # Explore neighboring states
    for neighbor in get_neighbors(state):
        if neighbor not in path_set:
            path.append(neighbor)
            path_set.add(neighbor)

            result = depth_limited_dfs(
                neighbor, goal, limit, path, path_set
            )

            if result is not None:
                return result

            # Backtrack
            path.pop()
            path_set.remove(neighbor)

    return None


def iddfs(start, goal):
    # Increase the depth limit gradually
    for limit in range(32):
        print("Searching at depth limit:", limit)

        path = [start]
        path_set = {start}

        result = depth_limited_dfs(
            start, goal, limit, path, path_set
        )

        if result is not None:
            return result

    return None


def display(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


# Get input from the user
start = tuple(map(int, input(
    "Enter initial state (9 numbers, use 0 for blank): "
).split()))

goal = tuple(map(int, input(
    "Enter goal state (9 numbers, use 0 for blank): "
).split()))

# Validate input
if (len(start) != 9 or len(goal) != 9
        or set(start) != set(range(9))
        or set(goal) != set(range(9))):
    print("Invalid input! Each state must contain 0 to 8 exactly once.")
else:
    solution = iddfs(start, goal)

    if solution is not None:
        print("\nSolution found!")

        for step, state in enumerate(solution):
            print("Step:", step)
            display(state)

        print("Total moves:", len(solution) - 1)
    else:
        print("No solution found within depth limit 31.")
```
