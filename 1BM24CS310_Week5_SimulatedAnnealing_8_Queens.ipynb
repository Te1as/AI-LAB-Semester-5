```python

import random
import math


def calculate_cost(state):
    n = len(state)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbor(state):
    neighbor = state.copy()

    # Choose two distinct columns and swap their rows
    i, j = random.sample(range(len(state)), 2)
    neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

    return neighbor, (i, j)

def print_grid(state):
    n = len(state)

    print("\nGrid:")
    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

def simulated_annealing():
    n = 8

    # Random initial state: one queen per column
    current = [0,1,2,3,4,5,6,7]
    current_cost = calculate_cost(current)

    temperature = 10.0
    cooling_rate = 0.995
    step = 0

    print("Initial state:", current)
    print("Initial cost:", current_cost)

    while temperature > 0.01:
        if current_cost == 0:
            print("\nGoal configuration reached!")
            print("Final state:", current)
            print("Final cost:", current_cost)
            print_grid(current)
            return

        neighbor, swapped_columns = get_neighbor(current)
        neighbor_cost = calculate_cost(neighbor)

        delta = neighbor_cost - current_cost

        # Accept improvements; sometimes accept worse states
        if delta < 0:
            accept = True
        else:
            probability = math.exp(-delta / temperature)
            accept = random.random() < probability

        print(f"\nStep {step}")
        print("Temperature:", round(temperature, 4))
        print("Current state:", current)
        print("Current cost:", current_cost)
        print("Swapped columns:", swapped_columns)
        print("Neighbor state:", neighbor)
        print("Neighbor cost:", neighbor_cost)
        print("Cost difference:", delta)
        print_grid(current)

        if accept:
            current = neighbor
            current_cost = neighbor_cost
            print("Move accepted")
        else:
            print("Move rejected")

        temperature *= cooling_rate
        step += 1

    print("\nTemperature too low.")
    print("Final state:", current)
    print("Final cost:", current_cost)

    if current_cost == 0:
        print("Goal configuration reached!")
    else:
        print("Solution not found. Try another run.")


simulated_annealing()
```
