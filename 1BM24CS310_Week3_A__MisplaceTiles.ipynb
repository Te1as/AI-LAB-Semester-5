```python

import heapq

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


def heuristic(state):
    misplaced = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            misplaced += 1

    return misplaced


def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])


def solve(start):
    start = tuple(start)

    # Priority queue: (f, counter, state, g, path)
    pq = []
    counter = 0

    g = 0
    h = heuristic(start)

    heapq.heappush(pq, (g + h, counter, start, g, [start]))

    best_g = {start: 0}

    while pq:
        f, _, state, g, path = heapq.heappop(pq)

        # Ignore outdated entries
        if g != best_g[state]:
            continue

        h = heuristic(state)

        if state == goal:
            print("Solution found!\n")

            for step, current in enumerate(path):
                g_value = step
                h_value = heuristic(current)

                print("Depth g(n):", g_value)
                print("Misplaced tiles h(n):", h_value)
                print("f(n) =", g_value + h_value)
                print_state(current)
                print()

            print("Total moves:", len(path) - 1)
            return

        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = [
            (-1, 0),  # Up
            (1, 0),   # Down
            (0, -1),  # Left
            (0, 1)    # Right
        ]

        for dr, dc in moves:
            nr = row + dr
            nc = col + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                new_zero = nr * 3 + nc

                next_state = list(state)
                next_state[zero], next_state[new_zero] = (
                    next_state[new_zero], next_state[zero]
                )
                next_state = tuple(next_state)

                new_g = g + 1

                if (next_state not in best_g or
                        new_g < best_g[next_state]):

                    best_g[next_state] = new_g
                    counter += 1

                    new_h = heuristic(next_state)

                    heapq.heappush(
                        pq,
                        (new_g + new_h, counter,
                         next_state, new_g, path + [next_state])
                    )

    print("No solution exists.")


start = list(map(int, input(
    "Enter the initial state (0 for blank): "
).split()))

if len(start) != 9 or set(start) != set(range(9)):
    print("Invalid input. Enter each number from 0 to 8 exactly once.")
else:
    solve(start)
```
