```python

from itertools import product

print(f"{'P':^3} {'Q':^3} {'R':^3} | {'Q->P':^6} {'P->~Q':^7} {'QvR':^5} | {'KB':^3}")
print("-" * 48)

models = []

for P, Q, R in product([True, False], repeat=3):

    # Individual sentences in KB
    s1 = (not Q) or P          # Q -> P
    s2 = (not P) or (not Q)   # P -> ~Q
    s3 = Q or R               # Q v R

    # Entire Knowledge Base
    KB = s1 and s2 and s3

    # Print truth table
    print(f"{int(P):^3} {int(Q):^3} {int(R):^3} | "
          f"{int(s1):^6} {int(s2):^7} {int(s3):^5} | "
          f"{int(KB):^3}")

    if KB:
        models.append((P, Q, R))

print("\nModels of P Q R where KB is TRUE:")
for model in models:
    print(tuple(int(x) for x in model))

# Check entailment
entails_R = all(R for P, Q, R in models)

entails_R_implies_P = all((not R) or P for P, Q, R in models)

entails_Q_implies_R = all((not Q) or R for P, Q, R in models)

print("\nEntailment Results:")
print("KB entails R:", entails_R)
print("KB entails R -> P:", entails_R_implies_P)
print("KB entails Q -> R:", entails_Q_implies_R)
```
