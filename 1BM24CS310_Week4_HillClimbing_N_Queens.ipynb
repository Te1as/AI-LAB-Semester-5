```python

def calculate_cost(state):
    n = len(state)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbors(state):
    n = len(state)
    neighbors = []

    # Swap pairs in ascending column-index order
    for i in range(n):
        for j in range(i + 1, n):
            neighbor = state.copy()
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append((i, j, neighbor))

    return neighbors

def print_grid(state):
    n = len(state)

    print("\nGrid:")
    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

def hill_climbing(initial_state):
    current = initial_state.copy()
    step = 0

    while True:
        current_cost = calculate_cost(current)

        print("\nStep:", step)
        print("Current state:", current)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("Goal configuration reached!")
            print_grid(current)
            return

        neighbors = get_neighbors(current)
        best_state = None
        best_cost = current_cost
        best_swap = None

        print("\nNeighbor costs:")

        for i, j, neighbor in neighbors:
            cost = calculate_cost(neighbor)
            print(f"Swap columns {i} and {j}: {neighbor}, cost = {cost}")

            # Strictly lower cost; ties retain the first swap
            if cost < best_cost:
                best_cost = cost
                best_state = neighbor
                best_swap = (i, j)

        if best_state is None:
            print("Local minimum reached.")
            return

        print("Chosen swap:", best_swap)
        print("Next state:", best_state)
        print("Next cost:", best_cost)
        print_grid(current)
        current = best_state
        step += 1


initial_state = [3, 1, 2, 0]
hill_climbing(initial_state)
```
